```python
%cd /kaggle/working/Rodinia/cuda/bfs

from pathlib import Path

path = Path("bfs.cu")
lines = path.read_text().splitlines(keepends=True)

# Make sure we are modifying the original, unmodified structure
if "cudaEvent_t start, stop_event;" in "".join(lines):
    raise RuntimeError("Timing code already exists. Do not run this cell again.")

# 1. Insert CUDA event declarations after: bool stop;
for i, line in enumerate(lines):
    if line.strip() == "bool stop;":
        lines[i+1:i+1] = [
            "\n",
            "\t\tcudaEvent_t start, stop_event;\n",
            "\t\tcudaEventCreate(&start);\n",
            "\t\tcudaEventCreate(&stop_event);\n"
        ]
        break
else:
    raise RuntimeError("Could not find 'bool stop;'.")

# 2. Insert start event immediately before the do loop
for i, line in enumerate(lines):
    if line.strip() == "do":
        lines[i:i] = [
            "\t\tcudaEventRecord(start);\n"
        ]
        break
else:
    raise RuntimeError("Could not find the BFS 'do' loop.")

# 3. Find the while(stop); and insert timing code after it
for i, line in enumerate(lines):
    if line.strip() == "while(stop);":
        lines[i+1:i+1] = [
            "\n",
            "\t\tcudaEventRecord(stop_event);\n",
            "\t\tcudaEventSynchronize(stop_event);\n",
            "\n",
            "\t\tfloat kernel_time_ms = 0.0f;\n",
            "\t\tcudaEventElapsedTime(&kernel_time_ms, start, stop_event);\n",
            "\n",
            "\t\tprintf(\"GPU Kernel Time: %.6f ms\\n\", kernel_time_ms);\n",
            "\n",
            "\t\tcudaEventDestroy(start);\n",
            "\t\tcudaEventDestroy(stop_event);\n"
        ]
        break
else:
    raise RuntimeError("Could not find 'while(stop);'.")

path.write_text("".join(lines))

print("CUDA timing code inserted successfully.")
```

```python
# ============================================================
# RODINIA CPU-GPU AUTOMATED BENCHMARK SYSTEM
# ============================================================

import os
import re
import subprocess
import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# CONFIGURATION
# ------------------------------------------------------------

RODINIA_ROOT = Path("/kaggle/working/Rodinia")

CPU_ROOT = RODINIA_ROOT / "openmp"
GPU_ROOT = RODINIA_ROOT / "cuda"
DATA_ROOT = RODINIA_ROOT / "data"

# Number of repeated measurements
N_RUNS = 6

# Output files
MASTER_DATASET = Path("/kaggle/working/rodinia_cpu_gpu_dataset.csv")
RAW_DATASET = Path("/kaggle/working/rodinia_raw_benchmark_results.csv")

print("Rodinia root:", RODINIA_ROOT)
print("CPU root:", CPU_ROOT)
print("GPU root:", GPU_ROOT)
print("Data root:", DATA_ROOT)
print("Number of runs per device:", N_RUNS)
```

```python
# ============================================================
# COMMAND EXECUTION FUNCTION
# ============================================================

def run_command(command, cwd):
    """
    Run a command inside a specified directory.

    Returns:
        stdout, stderr, return_code
    """

    result = subprocess.run(
        command,
        cwd=str(cwd),
        capture_output=True,
        text=True
    )

    return result.stdout, result.stderr, result.returncode
```

```python
# ============================================================
# CPU BENCHMARK FUNCTION
# ============================================================

def benchmark_cpu(
    workload,
    input_file,
    threads=4,
    n_runs=N_RUNS
):
    """
    Benchmark a Rodinia OpenMP workload.

    Expected command format:
        ./program <threads> <input_file>

    Returns a list containing raw CPU measurements.
    """

    cpu_dir = CPU_ROOT / workload
    executable = cpu_dir / workload

    if not executable.exists():
        raise FileNotFoundError(
            f"CPU executable not found:\n{executable}"
        )

    measurements = []

    print()
    print("=" * 60)
    print(f"CPU BENCHMARK: {workload}")
    print("=" * 60)

    for run in range(1, n_runs + 1):

        print(f"CPU Run {run}/{n_runs}")

        stdout, stderr, return_code = run_command(
            [
                f"./{workload}",
                str(threads),
                input_file
            ],
            cpu_dir
        )

        if return_code != 0:
            print("CPU execution failed.")
            print(stderr)
            raise RuntimeError(
                f"CPU benchmark failed for {workload}, run {run}"
            )

        # Rodinia BFS prints:
        # Compute time: 0.037446
        match = re.search(
            r"Compute time:\s*([0-9]*\.?[0-9]+)",
            stdout
        )

        if match is None:
            print(stdout)
            raise RuntimeError(
                f"Could not extract CPU execution time "
                f"for {workload}, run {run}"
            )

        cpu_time_seconds = float(match.group(1))
        cpu_time_ms = cpu_time_seconds * 1000.0

        measurements.append(cpu_time_ms)

        print(f"  CPU time: {cpu_time_ms:.6f} ms")

    print()
    print(
        f"CPU mean: {np.mean(measurements):.6f} ms"
    )

    return measurements
```

```python
# ============================================================
# GPU BENCHMARK FUNCTION
# ============================================================

def benchmark_gpu(
    workload,
    input_file,
    n_runs=N_RUNS
):
    """
    Benchmark a Rodinia CUDA workload.

    Extracts:
        GPU BFS/phase time
        H2D transfer time
        D2H transfer time
        Kernel iteration count
    """

    gpu_dir = GPU_ROOT / workload
    executable = gpu_dir / workload

    if not executable.exists():
        raise FileNotFoundError(
            f"GPU executable not found:\n{executable}"
        )

    measurements = []

    print()
    print("=" * 60)
    print(f"GPU BENCHMARK: {workload}")
    print("=" * 60)

    for run in range(1, n_runs + 1):

        print(f"GPU Run {run}/{n_runs}")

        stdout, stderr, return_code = run_command(
            [
                f"./{workload}",
                input_file
            ],
            gpu_dir
        )

        if return_code != 0:
            print(stdout)
            print(stderr)

            raise RuntimeError(
                f"GPU benchmark failed for {workload}, run {run}"
            )

        # ----------------------------------------------------
        # GPU phase time
        # ----------------------------------------------------

        phase_match = re.search(
            r"GPU Kernel Time:\s*([0-9]*\.?[0-9]+)\s*ms",
            stdout
        )

        # ----------------------------------------------------
        # H2D time
        # ----------------------------------------------------

        h2d_match = re.search(
            r"Initial Host to Device Time:\s*"
            r"([0-9]*\.?[0-9]+)\s*ms",
            stdout
        )

        # ----------------------------------------------------
        # D2H time
        # ----------------------------------------------------

        d2h_match = re.search(
            r"Final Device to Host Time:\s*"
            r"([0-9]*\.?[0-9]+)\s*ms",
            stdout
        )

        # ----------------------------------------------------
        # Kernel iterations
        # ----------------------------------------------------

        iteration_match = re.search(
            r"Kernel Executed\s+(\d+)\s+times",
            stdout
        )

        if phase_match is None:
            raise RuntimeError(
                f"Could not extract GPU phase time "
                f"for {workload}, run {run}"
            )

        if h2d_match is None:
            raise RuntimeError(
                f"Could not extract H2D time "
                f"for {workload}, run {run}"
            )

        if d2h_match is None:
            raise RuntimeError(
                f"Could not extract D2H time "
                f"for {workload}, run {run}"
            )

        if iteration_match is None:
            raise RuntimeError(
                f"Could not extract kernel iterations "
                f"for {workload}, run {run}"
            )

        phase_time_ms = float(phase_match.group(1))
        h2d_time_ms = float(h2d_match.group(1))
        d2h_time_ms = float(d2h_match.group(1))
        iterations = int(iteration_match.group(1))

        measurements.append({
            "gpu_phase_time_ms": phase_time_ms,
            "host_to_device_time_ms": h2d_time_ms,
            "device_to_host_time_ms": d2h_time_ms,
            "gpu_kernel_iterations": iterations
        })

        print(
            f"  GPU phase: {phase_time_ms:.6f} ms | "
            f"H2D: {h2d_time_ms:.6f} ms | "
            f"D2H: {d2h_time_ms:.6f} ms | "
            f"Iterations: {iterations}"
        )

    return measurements
```

```python
# ============================================================
# GPU HARDWARE INFORMATION
# ============================================================

def get_gpu_info():

    result = subprocess.run(
        [
            "nvidia-smi",
            "--query-gpu=name,memory.total",
            "--format=csv,noheader"
        ],
        capture_output=True,
        text=True
    )

    if result.returncode != 0:
        return "Unknown", np.nan

    lines = result.stdout.strip().splitlines()

    if len(lines) == 0:
        return "Unknown", np.nan

    # Use first GPU
    first_gpu = lines[0]

    parts = [x.strip() for x in first_gpu.split(",")]

    gpu_name = parts[0]

    memory_match = re.search(
        r"([0-9]+)",
        parts[1]
    )

    gpu_memory = (
        int(memory_match.group(1))
        if memory_match
        else np.nan
    )

    return gpu_name, gpu_memory


gpu_name, gpu_memory = get_gpu_info()

print("GPU:", gpu_name)
print("GPU memory:", gpu_memory, "MiB")
```

```python
# ============================================================
# COMPLETE WORKLOAD BENCHMARK
# ============================================================

def benchmark_workload(
    workload,
    cpu_input,
    gpu_input,
    input_size_mb=None,
    cpu_threads=4,
    n_runs=N_RUNS
):
    """
    Benchmark one Rodinia workload on both CPU and GPU.
    """

    print()
    print("#" * 70)
    print(f"# BENCHMARKING WORKLOAD: {workload}")
    print("#" * 70)

    # --------------------------------------------------------
    # CPU
    # --------------------------------------------------------

    cpu_times = benchmark_cpu(
        workload=workload,
        input_file=cpu_input,
        threads=cpu_threads,
        n_runs=n_runs
    )

    # --------------------------------------------------------
    # GPU
    # --------------------------------------------------------

    gpu_results = benchmark_gpu(
        workload=workload,
        input_file=gpu_input,
        n_runs=n_runs
    )

    # --------------------------------------------------------
    # Convert GPU measurements into arrays
    # --------------------------------------------------------

    gpu_phase_times = [
        x["gpu_phase_time_ms"]
        for x in gpu_results
    ]

    h2d_times = [
        x["host_to_device_time_ms"]
        for x in gpu_results
    ]

    d2h_times = [
        x["device_to_host_time_ms"]
        for x in gpu_results
    ]

    iterations = [
        x["gpu_kernel_iterations"]
        for x in gpu_results
    ]

    # --------------------------------------------------------
    # Calculate representative measurements
    # --------------------------------------------------------

    cpu_mean = np.mean(cpu_times)

    gpu_phase_mean = np.mean(gpu_phase_times)

    h2d_mean = np.mean(h2d_times)

    d2h_mean = np.mean(d2h_times)

    iteration_mean = np.mean(iterations)

    # Transfer-aware GPU path
    gpu_total = (
        gpu_phase_mean
        + h2d_mean
        + d2h_mean
    )

    # --------------------------------------------------------
    # Select measured device
    # --------------------------------------------------------

    preferred_device = (
        "GPU"
        if gpu_total < cpu_mean
        else "CPU"
    )

    # --------------------------------------------------------
    # Create dataset row
    # --------------------------------------------------------

    row = {
        "workload": workload,
        "input_file": Path(gpu_input).name,
        "input_size_mb": input_size_mb,

        "cpu_threads": cpu_threads,

        "gpu_type": gpu_name,
        "gpu_memory_mb": gpu_memory,

        "cpu_execution_time_ms": cpu_mean,

        "gpu_phase_time_ms": gpu_phase_mean,

        "host_to_device_time_ms": h2d_mean,

        "device_to_host_time_ms": d2h_mean,

        "gpu_total_path_time_ms": gpu_total,

        "gpu_kernel_iterations": iteration_mean,

        "preferred_device": preferred_device
    }

    # --------------------------------------------------------
    # Print summary
    # --------------------------------------------------------

    print()
    print("=" * 70)
    print(f"RESULT: {workload}")
    print("=" * 70)

    print(f"CPU mean time       : {cpu_mean:.6f} ms")
    print(f"GPU phase mean      : {gpu_phase_mean:.6f} ms")
    print(f"H2D mean            : {h2d_mean:.6f} ms")
    print(f"D2H mean            : {d2h_mean:.6f} ms")
    print(f"GPU total path      : {gpu_total:.6f} ms")
    print(f"Preferred device    : {preferred_device}")

    return row, cpu_times, gpu_results
```

```python
# ============================================================
# SAVE RAW BENCHMARK RESULTS
# ============================================================

def save_raw_results(
    workload,
    cpu_times,
    gpu_results
):

    rows = []

    for i in range(len(cpu_times)):

        gpu = gpu_results[i]

        rows.append({
            "workload": workload,
            "run": i + 1,

            "cpu_execution_time_ms":
                cpu_times[i],

            "gpu_phase_time_ms":
                gpu["gpu_phase_time_ms"],

            "host_to_device_time_ms":
                gpu["host_to_device_time_ms"],

            "device_to_host_time_ms":
                gpu["device_to_host_time_ms"],

            "gpu_kernel_iterations":
                gpu["gpu_kernel_iterations"]
        })

    new_data = pd.DataFrame(rows)

    if RAW_DATASET.exists():

        old_data = pd.read_csv(RAW_DATASET)

        combined = pd.concat(
            [old_data, new_data],
            ignore_index=True
        )

    else:

        combined = new_data

    combined.to_csv(
        RAW_DATASET,
        index=False
    )

    print()
    print("Raw results saved to:")
    print(RAW_DATASET)
```

```python
# ============================================================
# SAVE MASTER DATASET
# ============================================================

def save_master_row(row):

    new_row = pd.DataFrame([row])

    if MASTER_DATASET.exists():

        old_data = pd.read_csv(MASTER_DATASET)

        # Avoid accidental duplicate workload rows
        old_data = old_data[
            old_data["workload"] != row["workload"]
        ]

        combined = pd.concat(
            [old_data, new_row],
            ignore_index=True
        )

    else:

        combined = new_row

    combined.to_csv(
        MASTER_DATASET,
        index=False
    )

    print()
    print("Master dataset saved to:")
    print(MASTER_DATASET)

    return combined
```

```python
# ============================================================
# TEST COMPLETE SYSTEM ON BFS
# ============================================================

workload = "bfs"

input_file = "graph1MW_6.txt"

row, cpu_times, gpu_results = benchmark_workload(
    workload=workload,
    cpu_input=f"../../data/{workload}/{input_file}",
    gpu_input=f"../../data/{workload}/{input_file}",
    input_size_mb=62,
    cpu_threads=4,
    n_runs=6
)

save_raw_results(
    workload,
    cpu_times,
    gpu_results
)

dataset = save_master_row(row)

print()
print("CURRENT MASTER DATASET")
display(dataset)
```

```python
# ============================================================
# SECTION 55 — FIND TIMING CODE FOR ALL VALID WORKLOADS
# ============================================================

import re
from pathlib import Path

VALID_WORKLOADS = [
    "b+tree",
    "cfd",
    "heartwall",
    "hotspot",
    "hotspot3D",
    "kmeans",
    "leukocyte",
    "lud",
    "mummergpu",
    "myocyte",
    "nn"
]

TIMING_PATTERNS = [
    r"gettimeofday",
    r"getrusage",
    r"clock_gettime",
    r"omp_get_wtime",
    r"cudaEvent",
    r"cudaThreadSynchronize",
    r"cudaDeviceSynchronize",
    r"Start time",
    r"End time",
    r"Compute time",
    r"Execution time",
    r"Time",
    r"time",
    r"timer",
    r"Timer"
]

SOURCE_EXTENSIONS = {
    ".c",
    ".cc",
    ".cpp",
    ".cu",
    ".h",
    ".hpp"
}

for workload in VALID_WORKLOADS:

    print("\n" + "=" * 90)
    print(f"WORKLOAD: {workload}")
    print("=" * 90)

    for implementation, root in [
        ("CPU/OpenMP", CPU_ROOT / workload),
        ("GPU/CUDA", GPU_ROOT / workload)
    ]:

        print(f"\n--- {implementation} ---")

        matches_found = False

        if not root.exists():
            print("Directory not found.")
            continue

        source_files = [
            p for p in root.rglob("*")
            if p.is_file() and p.suffix.lower() in SOURCE_EXTENSIONS
        ]

        for source in sorted(source_files):

            try:
                text_content = source.read_text(
                    errors="ignore"
                )
            except Exception:
                continue

            lines = text_content.splitlines()

            matching_lines = []

            for line_no, line in enumerate(lines, start=1):

                line_lower = line.lower()

                if any(
                    re.search(pattern, line_lower)
                    for pattern in TIMING_PATTERNS
                ):
                    matching_lines.append(
                        (line_no, line.strip())
                    )

            if matching_lines:

                matches_found = True

                print(f"\nFile: {source.name}")

                # Show maximum 12 useful timing-related lines
                for line_no, line in matching_lines[:12]:
                    print(f"  {line_no:4d}: {line}")

        if not matches_found:
            print("No obvious timing code found.")
```

```python
# ============================================================
# SECTION 56 — SHOW MAKEFILES FOR VALID WORKLOADS
# ============================================================

for workload in VALID_WORKLOADS:

    print("\n" + "=" * 90)
    print(f"WORKLOAD: {workload}")
    print("=" * 90)

    for implementation, root in [
        ("CPU/OpenMP", CPU_ROOT / workload),
        ("GPU/CUDA", GPU_ROOT / workload)
    ]:

        print(f"\n--- {implementation} ---")

        makefiles = [
            p for p in root.iterdir()
            if p.is_file()
            and (
                p.name.lower() == "makefile"
                or p.name.lower() == "makefile_nvidia"
                or p.name.lower() == "makefile_nvidia"
            )
        ]

        if not makefiles:
            print("No Makefile found.")
            continue

        for makefile in makefiles:

            print(f"\n### {makefile.name}")

            text_content = makefile.read_text(
                errors="ignore"
            )

            # Print only useful compiler/build lines
            for line in text_content.splitlines():

                stripped = line.strip()

                if (
                    stripped.startswith("CC")
                    or stripped.startswith("CXX")
                    or stripped.startswith("NVCC")
                    or "gcc" in stripped
                    or "g++" in stripped
                    or "nvcc" in stripped
                    or stripped.startswith("all:")
                ):
                    print(stripped)
```

```python
# ============================================================
# SECTION 58 — GENERIC 6-RUN BENCHMARK ENGINE
# ============================================================

def run_command(command, cwd):
    """
    Execute one Rodinia command and capture stdout/stderr.
    """

    start = time.perf_counter()

    result = subprocess.run(
        command,
        cwd=str(cwd),
        shell=True,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    end = time.perf_counter()

    wall_time_ms = (end - start) * 1000.0

    return {
        "returncode": result.returncode,
        "stdout": result.stdout,
        "wall_time_ms": wall_time_ms
    }


def extract_value(text, regex_list):
    """
    Try multiple regex patterns.
    Return first successful numeric value.
    """

    for pattern in regex_list:

        match = re.search(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        if match:

            try:
                return float(match.group(1))
            except:
                pass

    return None


def extract_all_values(text, regex_list):
    """
    Return all numeric values matching supplied patterns.
    """

    values = []

    for pattern in regex_list:

        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        for match in matches:

            try:
                values.append(float(match))
            except:
                pass

    return values


def calculate_statistics(values):

    values = [
        float(x)
        for x in values
        if x is not None
        and math.isfinite(float(x))
    ]

    if not values:
        return {
            "mean": None,
            "median": None,
            "std": None,
            "min": None,
            "max": None
        }

    return {
        "mean": statistics.mean(values),
        "median": statistics.median(values),
        "std": statistics.stdev(values)
                if len(values) > 1 else 0.0,
        "min": min(values),
        "max": max(values)
    }


def get_gpu_info():

    try:

        command = (
            "nvidia-smi "
            "--query-gpu=name,memory.total "
            "--format=csv,noheader"
        )

        result = subprocess.run(
            command,
            shell=True,
            stdout=subprocess.PIPE,
            stderr=subprocess.STDOUT,
            text=True
        )

        lines = [
            line.strip()
            for line in result.stdout.splitlines()
            if line.strip()
        ]

        if lines:

            first_gpu = lines[0]

            parts = [
                x.strip()
                for x in first_gpu.split(",")
            ]

            gpu_name = parts[0]
            gpu_memory = parts[1]

            memory_match = re.search(
                r"([0-9.]+)",
                gpu_memory
            )

            gpu_memory_mb = (
                float(memory_match.group(1))
                if memory_match
                else None
            )

            return gpu_name, gpu_memory_mb

    except Exception:
        pass

    return None, None


def calculate_input_size_mb(input_files):

    total_bytes = 0

    for file_path in input_files:

        path = Path(file_path)

        if path.exists():

            total_bytes += path.stat().st_size

    return total_bytes / (1024 * 1024)


def benchmark_cpu(
    workload,
    config,
    n_runs=N_RUNS
):

    print("\n" + "-" * 70)
    print(f"CPU BENCHMARK: {workload}")
    print("-" * 70)

    times = []
    wall_times = []

    failed_runs = 0

    for run_number in range(1, n_runs + 1):

        print(
            f"CPU run {run_number}/{n_runs}...",
            end=" ",
            flush=True
        )

        result = run_command(
            config["command"],
            config["cwd"]
        )

        wall_times.append(
            result["wall_time_ms"]
        )

        if result["returncode"] != 0:

            failed_runs += 1

            print(
                f"FAILED (return code "
                f"{result['returncode']})"
            )

            print(result["stdout"][-1000:])

            continue

        measured = extract_value(
            result["stdout"],
            config.get("time_regex", [])
        )

        if measured is None:

            # Fall back to wall-clock time
            # but explicitly mark it.
            measured = result["wall_time_ms"]

            print(
                f"OK | wall={measured:.3f} ms "
                f"(application timing not detected)"
            )

        else:

            unit = config.get(
                "time_unit",
                "milliseconds"
            )

            if unit == "seconds":
                measured *= 1000.0

            print(
                f"OK | measured={measured:.3f} ms "
                f"| wall={result['wall_time_ms']:.3f} ms"
            )

        times.append(measured)

    stats = calculate_statistics(times)

    return {
        "times_ms": times,
        "wall_times_ms": wall_times,
        "statistics": stats,
        "failed_runs": failed_runs
    }


def benchmark_gpu(
    workload,
    config,
    n_runs=N_RUNS
):

    print("\n" + "-" * 70)
    print(f"GPU BENCHMARK: {workload}")
    print("-" * 70)

    phase_times = []
    h2d_times = []
    d2h_times = []
    iterations = []
    wall_times = []

    failed_runs = 0

    for run_number in range(1, n_runs + 1):

        print(
            f"GPU run {run_number}/{n_runs}...",
            end=" ",
            flush=True
        )

        result = run_command(
            config["command"],
            config["cwd"]
        )

        wall_times.append(
            result["wall_time_ms"]
        )

        if result["returncode"] != 0:

            failed_runs += 1

            print(
                f"FAILED (return code "
                f"{result['returncode']})"
            )

            print(result["stdout"][-1000:])

            continue

        phase = extract_value(
            result["stdout"],
            config.get("time_regex", [])
        )

        h2d = extract_value(
            result["stdout"],
            config.get("h2d_regex", [])
        )

        d2h = extract_value(
            result["stdout"],
            config.get("d2h_regex", [])
        )

        iteration = extract_value(
            result["stdout"],
            config.get("iterations_regex", [])
        )

        if phase is not None:

            unit = config.get(
                "time_unit",
                "milliseconds"
            )

            if unit == "seconds":
                phase *= 1000.0

            phase_times.append(phase)

        if h2d is not None:
            h2d_times.append(h2d)

        if d2h is not None:
            d2h_times.append(d2h)

        if iteration is not None:
            iterations.append(iteration)

        if phase is not None:

            print(
                f"OK | GPU phase={phase:.3f} ms"
            )

        else:

            # No application-level GPU timing found.
            # Keep wall-clock measurement separately.
            print(
                f"OK | GPU wall={result['wall_time_ms']:.3f} ms "
                f"(GPU phase timing not detected)"
            )

    phase_stats = calculate_statistics(
        phase_times
    )

    h2d_stats = calculate_statistics(
        h2d_times
    )

    d2h_stats = calculate_statistics(
        d2h_times
    )

    iteration_stats = calculate_statistics(
        iterations
    )

    wall_stats = calculate_statistics(
        wall_times
    )

    return {
        "phase_times_ms": phase_times,
        "h2d_times_ms": h2d_times,
        "d2h_times_ms": d2h_times,
        "iterations": iterations,
        "wall_times_ms": wall_times,

        "phase_statistics": phase_stats,
        "h2d_statistics": h2d_stats,
        "d2h_statistics": d2h_stats,
        "iteration_statistics": iteration_stats,
        "wall_statistics": wall_stats,

        "failed_runs": failed_runs
    }
```

```python
# ============================================================
# SECTION 59 — COMPLETE WORKLOAD BENCHMARK
# ============================================================

def benchmark_workload(
    workload,
    n_runs=N_RUNS
):

    if workload not in WORKLOAD_CONFIG:

        raise ValueError(
            f"{workload} is not configured."
        )

    config = WORKLOAD_CONFIG[workload]

    print("\n")
    print("=" * 80)
    print(f"BENCHMARKING: {workload}")
    print("=" * 80)

    # --------------------------------------------------------
    # Check directories
    # --------------------------------------------------------

    if not config["cpu"]["cwd"].exists():
        raise FileNotFoundError(
            f"CPU directory missing: "
            f"{config['cpu']['cwd']}"
        )

    if not config["gpu"]["cwd"].exists():
        raise FileNotFoundError(
            f"GPU directory missing: "
            f"{config['gpu']['cwd']}"
        )

    # --------------------------------------------------------
    # Input size
    # --------------------------------------------------------

    input_size_mb = calculate_input_size_mb(
        config.get("input_files", [])
    )

    # --------------------------------------------------------
    # GPU information
    # --------------------------------------------------------

    gpu_name, gpu_memory_mb = get_gpu_info()

    # --------------------------------------------------------
    # CPU
    # --------------------------------------------------------

    cpu_result = benchmark_cpu(
        workload,
        config["cpu"],
        n_runs
    )

    # --------------------------------------------------------
    # GPU
    # --------------------------------------------------------

    gpu_result = benchmark_gpu(
        workload,
        config["gpu"],
        n_runs
    )

    # --------------------------------------------------------
    # Extract means
    # --------------------------------------------------------

    cpu_mean = cpu_result[
        "statistics"
    ]["mean"]

    gpu_phase_mean = gpu_result[
        "phase_statistics"
    ]["mean"]

    h2d_mean = gpu_result[
        "h2d_statistics"
    ]["mean"]

    d2h_mean = gpu_result[
        "d2h_statistics"
    ]["mean"]

    gpu_wall_mean = gpu_result[
        "wall_statistics"
    ]["mean"]

    # --------------------------------------------------------
    # GPU total path
    #
    # Only calculate this when the GPU phase and both
    # transfer measurements are actually available.
    # --------------------------------------------------------

    if (
        gpu_phase_mean is not None
        and h2d_mean is not None
        and d2h_mean is not None
    ):

        gpu_total_path = (
            gpu_phase_mean
            + h2d_mean
            + d2h_mean
        )

    else:

        gpu_total_path = None

    # --------------------------------------------------------
    # Preferred device
    #
    # IMPORTANT:
    # This is a measured benchmark label.
    # It is NOT yet the ML prediction.
    # --------------------------------------------------------

    if (
        cpu_mean is not None
        and gpu_total_path is not None
    ):

        if gpu_total_path < cpu_mean:
            preferred_device = "GPU"
        else:
            preferred_device = "CPU"

    elif (
        cpu_mean is not None
        and gpu_wall_mean is not None
    ):

        if gpu_wall_mean < cpu_mean:
            preferred_device = "GPU"
        else:
            preferred_device = "CPU"

    else:

        preferred_device = "UNKNOWN"

    # --------------------------------------------------------
    # Result
    # --------------------------------------------------------

    row = {

        "workload": workload,

        "input_size_mb": round(
            input_size_mb,
            4
        ),

        "gpu_type": gpu_name,

        "gpu_memory_mb": gpu_memory_mb,

        "cpu_execution_time_ms":
            cpu_mean,

        "cpu_execution_median_ms":
            cpu_result["statistics"]["median"],

        "cpu_execution_std_ms":
            cpu_result["statistics"]["std"],

        "gpu_phase_time_ms":
            gpu_phase_mean,

        "gpu_phase_median_ms":
            gpu_result[
                "phase_statistics"
            ]["median"],

        "gpu_phase_std_ms":
            gpu_result[
                "phase_statistics"
            ]["std"],

        "host_to_device_time_ms":
            h2d_mean,

        "device_to_host_time_ms":
            d2h_mean,

        "gpu_wall_time_ms":
            gpu_wall_mean,

        "gpu_total_path_time_ms":
            gpu_total_path,

        "gpu_kernel_iterations":
            gpu_result[
                "iteration_statistics"
            ]["mean"],

        "preferred_device":
            preferred_device,

        "cpu_failed_runs":
            cpu_result["failed_runs"],

        "gpu_failed_runs":
            gpu_result["failed_runs"]
    }

    return row, cpu_result, gpu_result
```

```python
# ============================================================
# SECTION 60 — DATASET STORAGE
# ============================================================

def save_benchmark_row(row):

    # --------------------------------------------------------
    # Convert row to DataFrame
    # --------------------------------------------------------

    import pandas as pd

    new_row = pd.DataFrame([row])

    # --------------------------------------------------------
    # If dataset already exists, append/update workload
    # --------------------------------------------------------

    if MASTER_DATASET.exists():

        existing = pd.read_csv(
            MASTER_DATASET
        )

        # Remove previous result for the same workload
        existing = existing[
            existing["workload"]
            != row["workload"]
        ]

        combined = pd.concat(
            [existing, new_row],
            ignore_index=True
        )

    else:

        combined = new_row

    # --------------------------------------------------------
    # Save
    # --------------------------------------------------------

    combined.to_csv(
        MASTER_DATASET,
        index=False
    )

    return combined


def save_raw_run_results(
    workload,
    cpu_result,
    gpu_result
):

    import pandas as pd

    rows = []

    max_runs = max(
        len(cpu_result["times_ms"]),
        len(gpu_result["phase_times_ms"]),
        len(gpu_result["wall_times_ms"])
    )

    for i in range(max_runs):

        row = {
            "workload": workload,
            "run_number": i + 1,

            "cpu_time_ms": (
                cpu_result["times_ms"][i]
                if i < len(
                    cpu_result["times_ms"]
                )
                else None
            ),

            "cpu_wall_time_ms": (
                cpu_result["wall_times_ms"][i]
                if i < len(
                    cpu_result["wall_times_ms"]
                )
                else None
            ),

            "gpu_phase_time_ms": (
                gpu_result["phase_times_ms"][i]
                if i < len(
                    gpu_result["phase_times_ms"]
                )
                else None
            ),

            "h2d_time_ms": (
                gpu_result["h2d_times_ms"][i]
                if i < len(
                    gpu_result["h2d_times_ms"]
                )
                else None
            ),

            "d2h_time_ms": (
                gpu_result["d2h_times_ms"][i]
                if i < len(
                    gpu_result["d2h_times_ms"]
                )
                else None
            ),

            "gpu_wall_time_ms": (
                gpu_result["wall_times_ms"][i]
                if i < len(
                    gpu_result["wall_times_ms"]
                )
                else None
            ),

            "gpu_kernel_iterations": (
                gpu_result["iterations"][i]
                if i < len(
                    gpu_result["iterations"]
                )
                else None
            )
        }

        rows.append(row)

    new_raw = pd.DataFrame(rows)

    if RAW_DATASET.exists():

        existing_raw = pd.read_csv(
            RAW_DATASET
        )

        existing_raw = existing_raw[
            existing_raw["workload"]
            != workload
        ]

        combined_raw = pd.concat(
            [existing_raw, new_raw],
            ignore_index=True
        )

    else:

        combined_raw = new_raw

    combined_raw.to_csv(
        RAW_DATASET,
        index=False
    )

    return combined_raw
```

```python
# ============================================================
# SECTION 66 — CORRECTED CPU BENCHMARK
# ============================================================

def benchmark_cpu(
    workload,
    config,
    n_runs=N_RUNS
):

    print("\n" + "-" * 70)
    print(f"CPU BENCHMARK: {workload}")
    print("-" * 70)

    times = []
    wall_times = []

    failed_runs = 0
    parse_failures = 0

    for run_number in range(1, n_runs + 1):

        print(
            f"CPU run {run_number}/{n_runs}...",
            end=" ",
            flush=True
        )

        result = run_command(
            config["command"],
            config["cwd"]
        )

        wall_times.append(
            result["wall_time_ms"]
        )

        if result["returncode"] != 0:

            failed_runs += 1

            print(
                f"FAILED "
                f"(return code {result['returncode']})"
            )

            print(result["stdout"][-1000:])

            continue

        measured = extract_value(
            result["stdout"],
            config.get("time_regex", [])
        )

        if measured is None:

            parse_failures += 1

            print(
                "FAILED TO PARSE APPLICATION TIMING"
            )

            print(
                "Output:",
                result["stdout"][-500:]
            )

            continue

        unit = config.get(
            "time_unit",
            "milliseconds"
        )

        if unit == "seconds":
            measured *= 1000.0

        times.append(measured)

        print(
            f"OK | application time="
            f"{measured:.6f} ms"
        )

    stats = calculate_statistics(times)

    return {
        "times_ms": times,
        "wall_times_ms": wall_times,

        "statistics": stats,

        "failed_runs": failed_runs,

        "parse_failures": parse_failures
    }
```

```python
# ============================================================
# SECTION 67 — CORRECTED GPU BENCHMARK
# ============================================================

def benchmark_gpu(
    workload,
    config,
    n_runs=N_RUNS
):

    print("\n" + "-" * 70)
    print(f"GPU BENCHMARK: {workload}")
    print("-" * 70)

    phase_times = []
    h2d_times = []
    d2h_times = []
    iterations = []
    wall_times = []

    failed_runs = 0
    phase_parse_failures = 0

    for run_number in range(1, n_runs + 1):

        print(
            f"GPU run {run_number}/{n_runs}...",
            end=" ",
            flush=True
        )

        result = run_command(
            config["command"],
            config["cwd"]
        )

        wall_times.append(
            result["wall_time_ms"]
        )

        if result["returncode"] != 0:

            failed_runs += 1

            print(
                f"FAILED "
                f"(return code {result['returncode']})"
            )

            print(result["stdout"][-1000:])

            continue

        phase = extract_value(
            result["stdout"],
            config.get("time_regex", [])
        )

        h2d = extract_value(
            result["stdout"],
            config.get("h2d_regex", [])
        )

        d2h = extract_value(
            result["stdout"],
            config.get("d2h_regex", [])
        )

        iteration = extract_value(
            result["stdout"],
            config.get("iterations_regex", [])
        )

        if phase is None:

            phase_parse_failures += 1

            print(
                "FAILED TO PARSE GPU PHASE TIME"
            )

            print(
                "Output:",
                result["stdout"][-500:]
            )

            continue

        unit = config.get(
            "time_unit",
            "milliseconds"
        )

        if unit == "seconds":
            phase *= 1000.0

        phase_times.append(phase)

        if h2d is not None:
            h2d_times.append(h2d)

        if d2h is not None:
            d2h_times.append(d2h)

        if iteration is not None:
            iterations.append(iteration)

        print(
            f"OK | GPU phase="
            f"{phase:.6f} ms"
            f" | H2D="
            f"{h2d if h2d is not None else 'N/A'}"
            f" ms"
            f" | D2H="
            f"{d2h if d2h is not None else 'N/A'}"
            f" ms"
        )

    return {

        "phase_times_ms": phase_times,

        "h2d_times_ms": h2d_times,

        "d2h_times_ms": d2h_times,

        "iterations": iterations,

        "wall_times_ms": wall_times,

        "phase_statistics":
            calculate_statistics(phase_times),

        "h2d_statistics":
            calculate_statistics(h2d_times),

        "d2h_statistics":
            calculate_statistics(d2h_times),

        "iteration_statistics":
            calculate_statistics(iterations),

        "wall_statistics":
            calculate_statistics(wall_times),

        "failed_runs":
            failed_runs,

        "phase_parse_failures":
            phase_parse_failures
    }
```

```python
# ============================================================
# SECTION 69 — CORRECT RAW DATASET STORAGE
# ============================================================

def save_raw_run_results(
    workload,
    cpu_result,
    gpu_result
):

    import pandas as pd

    rows = []

    for i in range(N_RUNS):

        row = {
            "workload": workload,

            "run_number": i + 1,

            "cpu_time_ms": (
                cpu_result["times_ms"][i]
                if i < len(cpu_result["times_ms"])
                else None
            ),

            "cpu_wall_time_ms": (
                cpu_result["wall_times_ms"][i]
                if i < len(cpu_result["wall_times_ms"])
                else None
            ),

            "gpu_phase_time_ms": (
                gpu_result["phase_times_ms"][i]
                if i < len(gpu_result["phase_times_ms"])
                else None
            ),

            "h2d_time_ms": (
                gpu_result["h2d_times_ms"][i]
                if i < len(gpu_result["h2d_times_ms"])
                else None
            ),

            "d2h_time_ms": (
                gpu_result["d2h_times_ms"][i]
                if i < len(gpu_result["d2h_times_ms"])
                else None
            ),

            "gpu_wall_time_ms": (
                gpu_result["wall_times_ms"][i]
                if i < len(gpu_result["wall_times_ms"])
                else None
            ),

            "gpu_kernel_iterations": (
                gpu_result["iterations"][i]
                if i < len(gpu_result["iterations"])
                else None
            )
        }

        rows.append(row)

    new_raw = pd.DataFrame(rows)

    # Fixed column order
    columns = [
        "workload",
        "run_number",
        "cpu_time_ms",
        "cpu_wall_time_ms",
        "gpu_phase_time_ms",
        "h2d_time_ms",
        "d2h_time_ms",
        "gpu_wall_time_ms",
        "gpu_kernel_iterations"
    ]

    new_raw = new_raw[columns]

    # If the workload already exists, replace it
    if RAW_DATASET.exists():

        existing = pd.read_csv(
            RAW_DATASET
        )

        existing = existing[
            existing["workload"] != workload
        ]

        combined = pd.concat(
            [existing, new_raw],
            ignore_index=True
        )

        combined = combined[columns]

    else:

        combined = new_raw

    combined.to_csv(
        RAW_DATASET,
        index=False
    )

    return combined
```

```python
# ============================================================
# SECTION 70 — CORRECT MASTER DATASET STORAGE
# ============================================================

def save_benchmark_row(row):

    import pandas as pd

    columns = [
        "workload",
        "input_size_mb",
        "gpu_type",
        "gpu_memory_mb",

        "cpu_execution_time_ms",
        "cpu_execution_median_ms",
        "cpu_execution_std_ms",

        "gpu_phase_time_ms",
        "gpu_phase_median_ms",
        "gpu_phase_std_ms",

        "host_to_device_time_ms",
        "device_to_host_time_ms",

        "gpu_wall_time_ms",

        "gpu_total_path_time_ms",

        "gpu_kernel_iterations",

        "preferred_device",

        "cpu_failed_runs",
        "gpu_failed_runs"
    ]

    new_row = pd.DataFrame(
        [row],
        columns=columns
    )

    if MASTER_DATASET.exists():

        existing = pd.read_csv(
            MASTER_DATASET
        )

        # Remove previous result for same workload
        existing = existing[
            existing["workload"] != row["workload"]
        ]

        combined = pd.concat(
            [existing, new_row],
            ignore_index=True
        )

    else:

        combined = new_row

    combined = combined[columns]

    combined.to_csv(
        MASTER_DATASET,
        index=False
    )

    return combined
```

```python
# ============================================================
# SECTION 71 — CORRECT MASTER ROW CREATION
# ============================================================

def create_master_row(
    workload,
    cpu_result,
    gpu_result
):

    config = WORKLOAD_CONFIG[workload]

    input_size_mb = calculate_input_size_mb(
        config.get("input_files", [])
    )

    gpu_name, gpu_memory_mb = get_gpu_info()

    cpu_stats = cpu_result["statistics"]

    gpu_phase_stats = gpu_result[
        "phase_statistics"
    ]

    h2d_stats = gpu_result[
        "h2d_statistics"
    ]

    d2h_stats = gpu_result[
        "d2h_statistics"
    ]

    iteration_stats = gpu_result[
        "iteration_statistics"
    ]

    gpu_wall_stats = gpu_result[
        "wall_statistics"
    ]

    cpu_mean = cpu_stats["mean"]
    gpu_phase_mean = gpu_phase_stats["mean"]
    h2d_mean = h2d_stats["mean"]
    d2h_mean = d2h_stats["mean"]

    # --------------------------------------------------------
    # GPU total transfer-aware path
    # --------------------------------------------------------

    if (
        gpu_phase_mean is not None
        and h2d_mean is not None
        and d2h_mean is not None
    ):

        gpu_total = (
            gpu_phase_mean
            + h2d_mean
            + d2h_mean
        )

    else:

        gpu_total = None

    # --------------------------------------------------------
    # Measured scheduling label
    # --------------------------------------------------------

    if (
        cpu_mean is not None
        and gpu_total is not None
    ):

        if gpu_total < cpu_mean:
            preferred_device = "GPU"
        else:
            preferred_device = "CPU"

    else:

        preferred_device = "UNKNOWN"

    return {

        "workload": workload,

        "input_size_mb":
            input_size_mb,

        "gpu_type":
            gpu_name,

        "gpu_memory_mb":
            gpu_memory_mb,

        "cpu_execution_time_ms":
            cpu_mean,

        "cpu_execution_median_ms":
            cpu_stats["median"],

        "cpu_execution_std_ms":
            cpu_stats["std"],

        "gpu_phase_time_ms":
            gpu_phase_mean,

        "gpu_phase_median_ms":
            gpu_phase_stats["median"],

        "gpu_phase_std_ms":
            gpu_phase_stats["std"],

        "host_to_device_time_ms":
            h2d_mean,

        "device_to_host_time_ms":
            d2h_mean,

        "gpu_wall_time_ms":
            gpu_wall_stats["mean"],

        "gpu_total_path_time_ms":
            gpu_total,

        "gpu_kernel_iterations":
            iteration_stats["mean"],

        "preferred_device":
            preferred_device,

        "cpu_failed_runs":
            cpu_result["failed_runs"],

        "gpu_failed_runs":
            gpu_result["failed_runs"]
    }
```

```python
# ============================================================
# RODINIA 3.1 — CFD CPU/GPU 6-RUN BENCHMARK
# Clean measurement pipeline
# ============================================================

import os
import re
import time
import subprocess
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. PATHS
# ------------------------------------------------------------
RODINIA_ROOT = "/kaggle/working/Rodinia"

CPU_DIR = f"{RODINIA_ROOT}/openmp/cfd"
GPU_DIR = f"{RODINIA_ROOT}/cuda/cfd"

DATA_FILE = f"{RODINIA_ROOT}/data/cfd/fvcorr.domn.193K"

N_RUNS = 6

RAW_CSV = "/kaggle/working/rodinia_cfd_raw_results.csv"
MASTER_CSV = "/kaggle/working/rodinia_cfd_master_dataset.csv"

# ------------------------------------------------------------
# 2. CHECK PATHS
# ------------------------------------------------------------
print("=" * 70)
print("CFD BENCHMARK")
print("=" * 70)

print("\nChecking paths...")

print("CPU directory :", CPU_DIR, os.path.exists(CPU_DIR))
print("GPU directory :", GPU_DIR, os.path.exists(GPU_DIR))
print("Input file    :", DATA_FILE, os.path.exists(DATA_FILE))

if not os.path.exists(DATA_FILE):
    raise FileNotFoundError(f"CFD input file not found: {DATA_FILE}")

# ------------------------------------------------------------
# 3. COMPILE CPU
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("COMPILING CPU VERSION")
print("=" * 70)

cpu_make = subprocess.run(
    ["make", "-j2"],
    cwd=CPU_DIR,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

print(cpu_make.stdout[-4000:])

# Find likely CPU executable
CPU_EXEC = None

for candidate in [
    "euler3d_cpu",
    "euler3d_cpu.exe"
]:
    path = os.path.join(CPU_DIR, candidate)
    if os.path.exists(path):
        CPU_EXEC = path
        break

if CPU_EXEC is None:
    raise FileNotFoundError(
        "CPU CFD executable was not found after compilation."
    )

print("CPU executable:", CPU_EXEC)

# ------------------------------------------------------------
# 4. COMPILE GPU
# ------------------------------------------------------------
print("\n" + "=" * 70)
print("COMPILING GPU VERSION")
print("=" * 70)

gpu_make = subprocess.run(
    ["make", "-j2"],
    cwd=GPU_DIR,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True
)

print(gpu_make.stdout[-4000:])

GPU_EXEC = None

for candidate in [
    "euler3d",
    "euler3d.exe"
]:
    path = os.path.join(GPU_DIR, candidate)
    if os.path.exists(path):
        GPU_EXEC = path
        break

if GPU_EXEC is None:
    raise FileNotFoundError(
        "GPU CFD executable was not found after compilation."
    )

print("GPU executable:", GPU_EXEC)

# ------------------------------------------------------------
# 5. RUN COMMAND FUNCTION
# ------------------------------------------------------------
def run_command(command, cwd):

    start = time.perf_counter()

    result = subprocess.run(
        command,
        cwd=cwd,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    end = time.perf_counter()

    wall_ms = (end - start) * 1000.0

    return result.returncode, wall_ms, result.stdout


# ------------------------------------------------------------
# 6. TIMING PARSERS
# ------------------------------------------------------------
def parse_cpu_time(output):

    patterns = [

        # Example:
        # Compute time: 123.456
        r"Compute\s*time\s*:\s*([0-9.]+)",

        # Example:
        # Execution time: 123.456 ms
        r"Execution\s*time\s*:\s*([0-9.]+)",

        # Example:
        # Time: 123.456
        r"\bTime\s*:\s*([0-9.]+)"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            output,
            re.IGNORECASE
        )

        if match:
            return float(match.group(1))

    return np.nan


def parse_gpu_phase(output):

    patterns = [

        r"GPU\s+Phase\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"GPU\s+Kernel\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"Kernel\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"GPU\s+Execution\s+Time\s*:\s*([0-9.]+)\s*ms"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            output,
            re.IGNORECASE
        )

        if match:
            return float(match.group(1))

    return np.nan


def parse_h2d(output):

    patterns = [

        r"Initial\s+Host\s+to\s+Device\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"Host\s+to\s+Device\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"H2D\s+Time\s*:\s*([0-9.]+)\s*ms"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            output,
            re.IGNORECASE
        )

        if match:
            return float(match.group(1))

    return np.nan


def parse_d2h(output):

    patterns = [

        r"Final\s+Device\s+to\s+Host\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"Device\s+to\s+Host\s+Time\s*:\s*([0-9.]+)\s*ms",

        r"D2H\s+Time\s*:\s*([0-9.]+)\s*ms"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            output,
            re.IGNORECASE
        )

        if match:
            return float(match.group(1))

    return np.nan


def parse_iterations(output):

    patterns = [

        r"Kernel\s+Executed\s+(\d+)\s+times",

        r"Iterations\s*:\s*(\d+)"
    ]

    for pattern in patterns:

        match = re.search(
            pattern,
            output,
            re.IGNORECASE
        )

        if match:
            return int(match.group(1))

    return np.nan


# ------------------------------------------------------------
# 7. COMMANDS
# ------------------------------------------------------------

CPU_COMMAND = [
    "./euler3d_cpu",
    "../../data/cfd/fvcorr.domn.193K"
]

GPU_COMMAND = [
    "./euler3d",
    "../../data/cfd/fvcorr.domn.193K"
]

# ------------------------------------------------------------
# 8. BENCHMARK
# ------------------------------------------------------------

records = []

for run in range(1, N_RUNS + 1):

    print("\n" + "-" * 70)
    print(f"RUN {run}/{N_RUNS}")
    print("-" * 70)

    # --------------------------------------------------------
    # CPU
    # --------------------------------------------------------
    print("\nCPU running...")

    cpu_return, cpu_wall, cpu_output = run_command(
        CPU_COMMAND,
        CPU_DIR
    )

    cpu_time = parse_cpu_time(cpu_output)

    if cpu_return != 0:
        print("CPU FAILED")
        print(cpu_output[-2000:])

    else:
        print("CPU return code :", cpu_return)
        print("CPU application time:", cpu_time, "ms")
        print("CPU wall time:", cpu_wall, "ms")

    # --------------------------------------------------------
    # GPU
    # --------------------------------------------------------
    print("\nGPU running...")

    gpu_return, gpu_wall, gpu_output = run_command(
        GPU_COMMAND,
        GPU_DIR
    )

    gpu_phase = parse_gpu_phase(gpu_output)
    h2d = parse_h2d(gpu_output)
    d2h = parse_d2h(gpu_output)
    iterations = parse_iterations(gpu_output)

    if gpu_return != 0:

        print("GPU FAILED")
        print(gpu_output[-3000:])

    else:

        print("GPU return code :", gpu_return)
        print("GPU phase time  :", gpu_phase, "ms")
        print("H2D time        :", h2d, "ms")
        print("D2H time        :", d2h, "ms")
        print("GPU wall time   :", gpu_wall, "ms")
        print("Iterations      :", iterations)

    # --------------------------------------------------------
    # STORE ONE COMPLETE RECORD
    # --------------------------------------------------------
    records.append({
        "workload": "cfd",
        "run_number": run,

        "cpu_time_ms": cpu_time,
        "cpu_wall_time_ms": cpu_wall,

        "gpu_phase_time_ms": gpu_phase,
        "h2d_time_ms": h2d,
        "d2h_time_ms": d2h,

        "gpu_wall_time_ms": gpu_wall,
        "gpu_kernel_iterations": iterations,

        "cpu_return_code": cpu_return,
        "gpu_return_code": gpu_return
    })

# ------------------------------------------------------------
# 9. RAW DATASET
# ------------------------------------------------------------

raw_df = pd.DataFrame(records)

raw_df.to_csv(
    RAW_CSV,
    index=False
)

print("\n" + "=" * 70)
print("RAW DATASET")
print("=" * 70)

display(raw_df)

# ------------------------------------------------------------
# 10. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TIMING VALIDATION")
print("=" * 70)

print(
    "CPU timing values available:",
    raw_df["cpu_time_ms"].notna().sum(),
    "/",
    N_RUNS
)

print(
    "GPU phase values available:",
    raw_df["gpu_phase_time_ms"].notna().sum(),
    "/",
    N_RUNS
)

print(
    "H2D timing values available:",
    raw_df["h2d_time_ms"].notna().sum(),
    "/",
    N_RUNS
)

print(
    "D2H timing values available:",
    raw_df["d2h_time_ms"].notna().sum(),
    "/",
    N_RUNS
)

# ------------------------------------------------------------
# 11. SUMMARY STATISTICS
# ------------------------------------------------------------

cpu_mean = raw_df["cpu_time_ms"].mean()
cpu_median = raw_df["cpu_time_ms"].median()
cpu_std = raw_df["cpu_time_ms"].std()

gpu_phase_mean = raw_df["gpu_phase_time_ms"].mean()
gpu_phase_median = raw_df["gpu_phase_time_ms"].median()
gpu_phase_std = raw_df["gpu_phase_time_ms"].std()

h2d_mean = raw_df["h2d_time_ms"].mean()
d2h_mean = raw_df["d2h_time_ms"].mean()

# Transfer-aware GPU execution path
if not np.isnan(gpu_phase_mean) and not np.isnan(h2d_mean) and not np.isnan(d2h_mean):

    gpu_total = (
        gpu_phase_mean
        + h2d_mean
        + d2h_mean
    )

else:

    gpu_total = np.nan

# ------------------------------------------------------------
# 12. DEVICE LABEL
# ------------------------------------------------------------

if not np.isnan(cpu_mean) and not np.isnan(gpu_total):

    if gpu_total < cpu_mean:
        preferred_device = "GPU"
    else:
        preferred_device = "CPU"

else:

    preferred_device = "UNKNOWN"

# ------------------------------------------------------------
# 13. MASTER DATASET
# ------------------------------------------------------------

master_row = {

    "workload": "cfd",

    "input_file": "fvcorr.domn.193K",

    "cpu_execution_time_ms": cpu_mean,
    "cpu_execution_median_ms": cpu_median,
    "cpu_execution_std_ms": cpu_std,

    "gpu_phase_time_ms": gpu_phase_mean,
    "gpu_phase_median_ms": gpu_phase_median,
    "gpu_phase_std_ms": gpu_phase_std,

    "host_to_device_time_ms": h2d_mean,
    "device_to_host_time_ms": d2h_mean,

    "gpu_total_path_time_ms": gpu_total,

    "gpu_kernel_iterations":
        raw_df["gpu_kernel_iterations"].median(),

    "preferred_device": preferred_device,

    "cpu_failed_runs":
        int((raw_df["cpu_return_code"] != 0).sum()),

    "gpu_failed_runs":
        int((raw_df["gpu_return_code"] != 0).sum())
}

master_df = pd.DataFrame([master_row])

master_df.to_csv(
    MASTER_CSV,
    index=False
)

# ------------------------------------------------------------
# 14. FINAL RESULT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CFD MASTER RESULT")
print("=" * 70)

display(master_df)

print("\nFiles created:")
print(RAW_CSV)
print(MASTER_CSV)

print("\n" + "=" * 70)
print("IMPORTANT")
print("=" * 70)

print("""
CPU execution time  = application-reported CPU computation time
GPU phase time      = application-reported GPU phase time, if instrumented
H2D/D2H              = measured transfer times, if instrumented
GPU total path       = GPU phase + H2D + D2H

GPU wall time is NOT used as GPU execution time.
If a timing parser returns NaN, the code does NOT replace it with wall time.
""")
```

```python
import subprocess
import re
import numpy as np
import pandas as pd

GPU_DIR = "/kaggle/working/Rodinia/cuda/cfd"
DATA_FILE = "../../data/cfd/fvcorr.domn.193K"

N_RUNS = 6
results = []

print("=" * 70)
print("CFD GPU — 6 RUNS")
print("=" * 70)

for run in range(1, N_RUNS + 1):

    result = subprocess.run(
        ["./euler3d", DATA_FILE],
        cwd=GPU_DIR,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    output = result.stdout

    # CFD prints:
    # 0.00054809 seconds per iteration
    match = re.search(
        r"([0-9]+(?:\.[0-9]+)?)\s+seconds\s+per\s+iteration",
        output,
        re.IGNORECASE
    )

    gpu_time_ms = np.nan

    if match:
        seconds_per_iteration = float(match.group(1))
        gpu_time_ms = seconds_per_iteration * 1000.0

    results.append({
        "workload": "cfd",
        "run_number": run,
        "gpu_execution_time_ms": gpu_time_ms,
        "return_code": result.returncode
    })

    print(
        f"Run {run}: "
        f"{gpu_time_ms:.6f} ms per iteration "
        f"| return code = {result.returncode}"
    )

gpu_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("CFD GPU RESULTS")
print("=" * 70)

display(gpu_df)

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

print(
    f"Mean   : {gpu_df['gpu_execution_time_ms'].mean():.6f} ms"
)

print(
    f"Median : {gpu_df['gpu_execution_time_ms'].median():.6f} ms"
)

print(
    f"Std    : {gpu_df['gpu_execution_time_ms'].std():.6f} ms"
)
```

```python
import subprocess
import re
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. Location of CFD CUDA program and input data
# ---------------------------------------------------------

GPU_DIR = "/kaggle/working/Rodinia/cuda/cfd"
DATA_FILE = "../../data/cfd/fvcorr.domn.193K"

# We want 6 repeated measurements
N_RUNS = 6

results = []

print("=" * 70)
print("CFD GPU — 6 RUNS")
print("=" * 70)

# ---------------------------------------------------------
# 2. Run CFD six times
# ---------------------------------------------------------

for run in range(1, N_RUNS + 1):

    result = subprocess.run(
        ["./euler3d", DATA_FILE],
        cwd=GPU_DIR,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )

    output = result.stdout

    # CFD prints a line like:
    #
    # 0.00054809 seconds per iteration
    #
    # We extract the number before
    # "seconds per iteration".

    match = re.search(
        r"([0-9]+(?:\.[0-9]+)?)\s+seconds\s+per\s+iteration",
        output,
        re.IGNORECASE
    )

    gpu_time_ms = np.nan

    if match:
        seconds_per_iteration = float(match.group(1))

        # Convert seconds to milliseconds
        gpu_time_ms = seconds_per_iteration * 1000.0

    # Store this run
    results.append({
        "workload": "cfd",
        "run_number": run,
        "gpu_execution_time_ms": gpu_time_ms,
        "return_code": result.returncode
    })

    print(
        f"Run {run}: "
        f"{gpu_time_ms:.6f} ms per iteration "
        f"| return code = {result.returncode}"
    )


# ---------------------------------------------------------
# 3. Create a table containing all 6 runs
# ---------------------------------------------------------

gpu_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("CFD GPU RESULTS")
print("=" * 70)

display(gpu_df)


# ---------------------------------------------------------
# 4. Calculate statistics
# ---------------------------------------------------------

print("\n" + "=" * 70)
print("SUMMARY")
print("=" * 70)

print(
    f"Mean   : "
    f"{gpu_df['gpu_execution_time_ms'].mean():.6f} ms"
)

print(
    f"Median : "
    f"{gpu_df['gpu_execution_time_ms'].median():.6f} ms"
)

print(
    f"Std    : "
    f"{gpu_df['gpu_execution_time_ms'].std():.6f} ms"
)
```
